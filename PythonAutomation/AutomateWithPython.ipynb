## Merge PDF

**Read multiple PDF files from the directory and merge them to a single PDF file**

### Import Libraries

Import `os` to list files & directories. Also, import [`pypdf`](https://pypi.org/project/pypdf/) to merge multiple PDF files into a single PDF file. ([Read Doc pypdf](https://pypdf.readthedocs.io/en/latest/))

```python
import os
from pypdf import PdfMerger, PdfWriter
```

### Set Directory Containing PDF Files

```python
import os
from pypdf import PdfWriter

def pdf_merger(input_dir, output_dir=os.getcwd()):
    for dir, dirname, files in os.walk(input_dir):
        output_file = files[0].split(" ")[0]
        merger = PdfWriter()
        for file in files:
            if file.endswith(".pdf"):
                merger.append(os.path.join(dir, file))
        merger.write(f"{output_dir}/{output_file}_merged.pdf")
        merger.close()
```

### Call `pdf_merger` Function

```python
input_dir = "/home/user/Desktop/MergeFiles"
output_dir = "/home/user/Desktop/MergeFiles"

pdf_merger(input_dir, output_dir)
```

## Add Page Number in PDF

Use `pypdf` to read & write PDF file and `reportlab` to add page numbers in PDF.

```python
from io import BytesIO
import os
from pypdf import PdfReader, PdfWriter
from reportlab.lib.pagesizes import A4, portrait
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfgen import canvas

kalimati_font = os.path.join(os.getcwd(), 'Kalimati.ttf')
pdfmetrics.registerFont(TTFont("Kalimati", kalimati_font))

# pdfmetrics.registerFont(TTFont("Kalimati", "/home/user/Desktop/MergeFiles/Kalimati.ttf"))

def add_page_numbers(input_pdf, output_pdf):
    # Read existing PDF to add page number
    reader = PdfReader(input_pdf)
    writer = PdfWriter()

    # Create a temporary PDF to add page numbers
    packet = BytesIO()
    c = canvas.Canvas(packet, pagesize=portrait(A4))
    

    # Loop through each page and add page number
    for page_num in range(len(reader.pages)):
        c.showPage()
        # Adjust the font and position of the page number
        c.setFont("Kalimati",7)
        c.drawString(300, 30, f"{page_num + 1}")
    c.setPageCompression(100)
    c.save()

    # Goto the start of the BytesIO object
    packet.seek(0)

    # Create a temporary PDF with page number
    page_number_pdf = PdfReader(packet)

    # Add page numbers to each PDF page
    for page_num in range(len(reader.pages)):
        page = reader.pages[page_num]
        packet_page = page_number_pdf.pages[page_num + 1]

        # Merge the page with page number overlay
        page.merge_page(packet_page)

        # Add merged page to the writer
        writer.add_page(page)

    # Write the final output PDF
    with open(output_pdf, "wb") as output_file:
        writer.write(output_file)

# Use Function
input_pdf = "/home/user/Desktop/MergeFiles/merged.pdf"
output_pdf = "output.pdf"
add_page_numbers(input_pdf, output_pdf)
```
